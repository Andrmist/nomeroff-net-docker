```python
import pandas as pd
import requests
```

```python
numberplates = pd.read_csv("numberplates.csv", dtype={
    "numberplate_id": str,
    "auto_id": 'Int64',
    "photo_index": 'Int64',
    "photo_id": 'Int64',
    "ordering": 'Int64',
    "partner_id": 'Int64'
})
```

```python
numberplates["numberplate_id"] = numberplates["numberplate_id"].apply(lambda x: x[2:])
numberplates.head()
```

```python
grouped = numberplates.groupby('auto_id').agg({
    'number': lambda x: x.iloc[0],
    'numberplate_url': lambda x: x.iloc[0],
    'image_url': lambda x: ','.join(x)
}).reset_index()
grouped.head()
```

```python
def numberplate_parse_old(numberplate_url):
    url = "http://dc.cherniy.net:2236/read?url="
    try:
        res = requests.get(f"{url}{numberplate_url}")
        data = res.json()
        return data["number_plates"][0]
    except Exception as e:
        print(e)
        return None

def numberplate_parse_new(numberplate_url):
    url = "http://localhost:3116/read?url="
    try:
        res = requests.get(f"{url}{numberplate_url}")
        data = res.json()
        return data["number_plates"][0]
    except Exception as e:
        print(e)
        return None
```

```python
grouped["old"] = grouped["numberplate_url"].apply(numberplate_parse_old)
grouped["new"] = grouped["image_url"].apply(numberplate_parse_new)
```

```python
grouped["new"] = grouped["image_url"].apply(numberplate_parse_new)
grouped.head(50)
```

```python
diff1 = grouped.query("number == new and number != old")
diff2 = grouped.query("number != new and number == old")
diff2
```

```python
grouped_fix = grouped.copy(deep=True)
grouped_fix.head()
```

```python
grouped_fix["old"] = grouped_fix["old"].apply(lambda x: "" if x is None else x)
grouped_fix["new"] = grouped_fix["new"].apply(lambda x: "" if x is None else x)
grouped_fix.head()
```

```python
old_count = grouped.query("number == old").shape[0]
new_count = grouped.query("number == new").shape[0]
total_count = grouped.shape[0]
failed = grouped.query("old.isnull() & new.isnull()").shape[0]
print(f"""old_count: {old_count}
new_count: {new_count}
total_count = {total_count}
failed = {failed}
total_photo_count = {numberplates.shape[0]}
delta = {new_count - old_count}
percentage = {(new_count - old_count) / new_count}
""")
```

```python
old_count_fix = grouped_fix.query("number == old").shape[0]
new_count_fix = grouped_fix.query("number == new").shape[0]
total_count_fix = grouped_fix.shape[0]
failed_fix = grouped_fix.query("old == '' and new == ''").shape[0]
print(f"""old_count: {old_count_fix}
new_count: {new_count_fix}
total_count = {total_count_fix}
failed = {failed_fix}
total_photo_count = {numberplates.shape[0]}
delta = {new_count_fix - old_count_fix}
percentage = {(new_count_fix - old_count_fix) / new_count_fix}
""")
```

```python
grouped.to_csv("result.csv", index=None)
```
